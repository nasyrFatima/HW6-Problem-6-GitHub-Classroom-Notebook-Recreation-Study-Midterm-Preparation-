**Pearson Correlation Feature Ranking**

**r_regression feature importance**

Each feature and target's linear correlation is calculated using this method.

Essential concept:

Positive score → direct correlation

Inverse relationship: negative score

Greater absolute value implies greater significance.

Absolute magnitude is used to sort features.

A bar plot maintains ranking while displaying signed data.

**Top 5 Pearson features**

The top five variables are listed.

The highest absolute correlations are the only ones chosen.

This aids in determining which predictors, under linear assumptions, have the most impact on the target.

**Mutual Information Ranking**

**mutual_info_regression**

Dependency is measured by mutual information, which goes beyond basic linear correlation.

In contrast to Pearson:

captures nonlinear effects.

Scores are never negative.

Stronger predictive relationships are indicated by higher values.

The features are arranged from highest to lowest score.

The final ranking is shown in a bar graph.

**Top 5 mutual information features**

Based on mutual information, the top five variables are listed.

Because nonlinear interactions are taken into account, this ranking might be different from Pearson.

**Random Forest Feature Importance**

**Random forest importance**

Training data is used to train a random forest regressor.

The significance of a feature is taken from:

feature_importances_

Interpretation:

importance is the amount that each tree contributes to lowering prediction error.

Stronger contributions are indicated by large values.

For reliable estimates, 500 trees are utilized.

**Top 5 random forest features**

The top five random forest significance scores are shown.

This frequently highlights variables that are utilized frequently throughout tree splits.

**Recursive Feature Elimination (RFE)**

**RFE with SVR**

Weaker variables are gradually eliminated by recursive feature removal.

The estimator is a linear SVR model.

Procedure:

fit model

Eliminate the least helpful feature

Continue until a ranking is achieved.

Stronger features are those with lower ranks.

The best feature is denoted by rank = 1.

**Top 5 RFE features**

Ranking output is used to display the top five features.

Strongest variables are indicated by the smallest ranking values.

**Overall comparison**

Different methods produce slightly different rankings because each measures importance differently:

Pearson focuses on linear correlation

Mutual information captures nonlinear dependence

Random forest measures contribution in tree models

RFE evaluates features through repeated elimination

Features that appear repeatedly near the top are likely consistently informative predictors.
